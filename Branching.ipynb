```python
# Question 4 -- 2

b = str(input("Enter a number to check for palindrome: "))
rev = ''
c = len(b) - 1

for i in b:
    rev += b[c]
    c -= 1

if b == rev:
    print("Is Palindrome");

else:
    print("Not a Palidrome");
```

```python
# Question 4 -- 3
d = str(input("Enter a number to check for palindrome: "))
summing = 0
e = len(d) - 1

for i in d:
    summing += int(d[e])
    e -= 1

num = int(summing)
input_intial = int(d)
if (input_intial % num == 0):
    print("Divisble");


else:
    print("Not Divisible")
```
