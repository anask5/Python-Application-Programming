```python
name1 = "Mobile Phone"
quantity1 = 5
price1 = 20000
year1 = 2020
```

```python
Product1 = {
    "name" : name1,
    "quantity" : quantity1,
    "price" : price1,
    "year" : year1
}
```

```python
name2 = "IPhone"
quantity2 = 12
price2 = 90000
year2 = 2023
```

```python
Product2 = {
    "name" : name2,
    "quantity" : quantity2,
    "price" : price2,
    "year" : year2
}
```

```python
Inventory = {
    "Product1" : Product1,
    "Product2" : Product2
}
```

```python
print(Inventory)
```

```python

print(Inventory["Product1"]["year"])
```

```python

print(Inventory["Product2"]["year"])
```

```python
del Inventory["Product1"]["year"]
del Inventory["Product2"]["year"]
```

```python
print(Inventory)
```
